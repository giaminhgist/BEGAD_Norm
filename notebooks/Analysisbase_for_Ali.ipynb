```python
import pandas as pd
import numpy as np
#import begatools
import os
```

```python
def processCSV(df):
    tasklst = []
    for r,row in df.iterrows():
        #if row['SaccadeIndex']>0 and row['Final']==True:# and row['HasNulls']==False:
        if row['FixationIndex']>0 and row['Final']==True and row['HasNulls']==False:
            tasklst.append(row)
    return pd.DataFrame(tasklst,columns=df.columns)


useColumns = ['PupilLeft','PupilRight','AbsoluteSaccadicDirection','RelativeSaccadicDirection','SaccadicAmplitude','GazeEventDuration','SacSpeed','DistanceLeft','DistanceRight','RecordingTimestamp','Group','SOB']

groups = ['CO', 'MCI', 'AD']

```

```python
rdf = pd.DataFrame()
folder = 'BEGAD_ETRA/20250305/'
listf = os.listdir(folder)
totalf = len(listf)
for f,file in enumerate(listf):
    if file.endswith('.csv'):
        print(f'Ingesting {file} ',end="")
        print(str(f)+' / '+str(totalf)+' files ingested ')
        thisdf = processCSV(pd.read_csv(folder+file))
        rdf = pd.concat([rdf,thisdf],ignore_index=0)
    else:
        totalf-=1

```

```python
fdf = rdf 
len(fdf) #let's check how many rows of data we have
```
